## Anomaly Detection & Time Series

1. What is Anomaly Detection? Explain its types (point, contextual, and
collective anomalies) with examples.?

Ans:Anomaly detection is the process of identifying data points or patterns that deviate significantly from normal or expected behavior within a dataset, often signaling a potential issue or a rare event. The three main types are point anomalies, which are individual instances that stand out alone; contextual anomalies, which are unusual given their specific circumstances but not otherwise; and collective anomalies, which are groups of related data points that together exhibit unusual behavior, even if individual points seem normal.

---Types of Anomalies:

1. Point Anomalies:

 - Definition: A data point is considered an anomaly if it is significantly different from the rest of the data.

 - Example:
In a bank’s transaction system, if the average daily transaction amount is around $500, but one transaction is $50,000, that transaction is a point anomaly.

 - In temperature readings, if most days in winter are around 5°C and one day records 30°C, that’s a point anomaly.

2. Contextual Anomalies (Conditional Anomalies):

 - Definition: A data point is anomalous in a specific context, but may not be anomalous in another. The context is usually defined by time, location, or surrounding data.

 - Example:
A temperature of 30°C in summer is normal, but the same 30°C in winter is a contextual anomaly.

 - A heart rate of 110 beats per minute during exercise is normal, but the same rate while sleeping is anomalous.

3. Collective Anomalies:

 - Definition: A collection (subsequence or group) of data points is anomalous when considered together, even though individual points may not look abnormal.

 - Example:
In network traffic, a sudden burst of requests within a short period may indicate a DDoS attack, even if each request looks normal on its own.

  - In stock markets, a series of small but consistent price drops over several days could signal insider trading or market manipulation.

2. Compare Isolation Forest, DBSCAN, and Local Outlier Factor in terms of
their approach and suitable use cases. ?

Ans:Isolation Forest identifies outliers by isolating them in trees, excels at detecting global anomalies in large, high-dimensional datasets, and is efficient but less effective at local outliers. DBSCAN, a density-based clustering algorithm, groups dense data points into clusters and labels sparse points as noise, making it ideal for noisy data where cluster shape isn't fixed. Local Outlier Factor (LOF) measures local density deviations, identifying outliers that are significantly less dense than their neighbors and suitable for high-dimensional data where global density assumptions fail.


1. Isolation Forest (IF)

 ----Approach:

 - Based on the idea that anomalies are “few and different.”

 - Uses random decision trees to isolate data points.

 - Anomalies are easier to isolate (fewer splits needed).

----Strengths:

 - Works well with high-dimensional data.

 - Scalable to large datasets.

 - Model-based, not distance-based → less affected by the curse of dimensionality.

----Weaknesses:

 - Assumes anomalies are rare and different in feature space.

 - May miss anomalies if data distribution is highly complex.

----Use Case Example:

 - Fraud detection in banking: Identifying unusual transactions in millions of records.

2. DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

----Approach:

 - Clustering algorithm that groups points into dense regions.

 - Points not belonging to any dense region are labeled as outliers.

---Strengths:

 - Finds anomalies as points in low-density regions.

 - Can detect clusters of arbitrary shape (not limited to spherical).

 - No need to specify the number of clusters.

----Weaknesses:

 - Struggles with high-dimensional data (distance measure loses meaning).

 - Requires careful tuning of parameters (eps, minPts).

----Use Case Example:

 - Geospatial anomaly detection: Identifying unusual GPS trajectories or outlier events in location data.

3. Local Outlier Factor (LOF)

-----Approach:

 - Density-based method that measures the local density of a point compared to its neighbors.

 - A point is anomalous if its density is much lower than that of its neighbors.

----Strengths:

 - Detects local anomalies (points anomalous only in their neighborhood).

 - Good for datasets where anomaly meaning is context-dependent.

-----Weaknesses:

 - Computationally expensive on large datasets.

 - Sensitive to parameter choice (k, neighborhood size).

----Use Case Example:

 - Network intrusion detection: A host with unusual traffic compared to other similar hosts.

3. What are the key components of a Time Series? Explain each with one
example.?


Ans:The four key components of a time series are Trend, Seasonal Variation, Cyclical Variation, and Irregular Variation (Noise). A trend is the long-term, upward or downward movement of the data. Seasonal variation is a predictable, repeating pattern within a year, like increased ice cream sales in the summer. Cyclical variation is a repeating pattern of fluctuations that occur over longer periods than a year, often tied to business cycles. Irregular variation, or noise, represents unpredictable, random fluctuations in the data.



---Key Components of a Time Series:

1. Trend:

 - Definition: The long-term movement or direction in the data (upward, downward, or flat) over time.

 - Example: The number of internet users worldwide shows a steadily increasing trend over the years.

2. Seasonality:

 - Definition: Regular, repeating patterns within a fixed period (daily, weekly, yearly, etc.).

 - Example: Ice cream sales peak in summer and drop in winter, showing yearly seasonality.

3. Cyclic Component:

 - Definition: Fluctuations that occur over longer, irregular periods (not strictly seasonal). Often tied to economic or business cycles.

 - Example: Stock market prices showing bull and bear markets every few years.

4. Irregular/Random Component (Noise):

 - Definition: Random variations or unexpected shocks in the data that cannot be explained by trend, seasonality, or cycles.

 - Example: A sudden drop in airline passengers due to COVID-19 lockdowns.

4. Define Stationary in time series. How can you test and transform a
non-stationary series into a stationary one?

Ans:A time series is stationary if its statistical properties, such as mean, variance, and autocorrelation, are constant over time, meaning the series is unaffected by the time at which it is viewed. To test for stationarity, you can use visualizations (like run sequence plots), analyze summary statistics of data partitions, or apply statistical tests like the Augmented Dickey-Fuller (ADF) test or the Kwiatkowski-Phillips-Schmidt-Shin (KPSS) test. Non-stationary series can be transformed into stationary ones through techniques such as differencing, log transformations (to stabilize variance), and detrending or seasonal decomposition to remove trends and seasonality.

A stationary time series is one whose statistical properties do not change over time.
This means:

 - Constant mean

 - Constant variance

 - Constant autocovariance (correlation structure does not depend on time, only on lag)

  - In simple terms: the data should "look the same" at any point in time.

-----Why is Stationarity Important:

 - Many time series models (like ARIMA) assume stationarity.

 - Non-stationary data can lead to spurious results in forecasting and correlation analysis.

----How to Test Stationarity:

 1.  Visual Inspection:

 - Plot the time series.

 - If it shows trend, changing variance, or seasonality, it’s likely non-stationary.

2. Statistical Tests:

 - Augmented Dickey-Fuller (ADF) Test

 - Null hypothesis (H₀): Series is non-stationary.

 - If p-value < 0.05 → reject H₀ → series is stationary.

 - KPSS Test (Kwiatkowski-Phillips-Schmidt-Shin)

 - Null hypothesis (H₀): Series is stationary.

 - If p-value < 0.05 → reject H₀ → series is non-stationary.

 - Often, both tests are used together for confirmation.

3. Rolling Statistics:

 - Compute rolling mean and variance.

 - If they change significantly over time → non-stationary.

```python
# 6. Dataset:
# ● NYC Taxi Fare Data
#● AirPassengers Dataset
#Question 6: Load a time series dataset (e.g., AirPassengers), plot the original series,
#and decompose it into trend, seasonality, and residual components.

import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# Load AirPassengers dataset (available in statsmodels)
from statsmodels.datasets import get_rdataset
data = get_rdataset("AirPassengers")  # monthly airline passenger data
df = data.data

# Convert to proper datetime format
df['time'] = pd.date_range(start='1949-01', periods=len(df), freq='M')
df.set_index('time', inplace=True)
df.rename(columns={'value': 'Passengers'}, inplace=True)

# Plot original series
plt.figure(figsize=(10,4))
plt.plot(df['Passengers'], label='AirPassengers')
plt.title("Original AirPassengers Time Series")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.legend()
plt.show()

# Decompose (Additive model is common for AirPassengers)
decomposition = seasonal_decompose(df['Passengers'], model='multiplicative', period=12)

# Plot decomposition
fig = decomposition.plot()
fig.set_size_inches(10,8)
plt.show()

```

```python
# 7. Apply Isolation Forest on a numerical dataset (e.g., NYC Taxi Fare) to
# detect anomalies. Visualize the anomalies on a 2D scatter plot.



import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# Load dataset (replace with actual file path if available)
# Example: df = pd.read_csv("nyc_taxi_fare.csv")
# For demonstration, let's assume we have only 'fare_amount' and 'trip_distance'
# (You can select more numerical columns as needed)

# Example dummy data (replace with real data)
data = {
    'fare_amount': [5, 7, 10, 12, 8, 50, 7, 9, 100, 6, 8, 11, 200],  # some extreme anomalies
    'trip_distance': [1.2, 1.5, 2.1, 2.3, 1.8, 15, 2.0, 2.4, 30, 1.3, 2.1, 2.6, 50]
}
df = pd.DataFrame(data)

# Apply Isolation Forest
clf = IsolationForest(contamination=0.1, random_state=42)
df['anomaly'] = clf.fit_predict(df[['fare_amount', 'trip_distance']])

# Anomalies are labeled as -1
anomalies = df[df['anomaly'] == -1]
normal = df[df['anomaly'] == 1]

# Scatter plot
plt.figure(figsize=(8,6))
plt.scatter(normal['trip_distance'], normal['fare_amount'],
            c='blue', label='Normal', alpha=0.7)
plt.scatter(anomalies['trip_distance'], anomalies['fare_amount'],
            c='red', label='Anomaly', alpha=0.9, marker='x', s=100)
plt.xlabel("Trip Distance (miles)")
plt.ylabel("Fare Amount ($)")
plt.title("Isolation Forest - NYC Taxi Fare Anomaly Detection")
plt.legend()
plt.show()

```

```python
# 8.  Train a SARIMA model on the monthly airline passengers dataset.
# Forecast the next 12 months and visualize the results.

import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.datasets import get_rdataset

# Load AirPassengers dataset (monthly airline passengers 1949–1960)
data = get_rdataset("AirPassengers").data
df = data.copy()
df['Month'] = pd.date_range(start='1949-01', periods=len(df), freq='M')
df.set_index('Month', inplace=True)
df.rename(columns={'value': 'Passengers'}, inplace=True)

# Train SARIMA model
# Parameters chosen based on ACF/PACF knowledge for this dataset
model = SARIMAX(df['Passengers'],
                order=(1,1,1),          # (p,d,q)
                seasonal_order=(1,1,1,12),  # (P,D,Q,m) with m=12 for yearly seasonality
                enforce_stationarity=False,
                enforce_invertibility=False)

results = model.fit(disp=False)

# Forecast next 12 months
forecast = results.get_forecast(steps=12)
forecast_ci = forecast.conf_int()

# Plot results
plt.figure(figsize=(10,6))
plt.plot(df['Passengers'], label='Observed')
plt.plot(forecast.predicted_mean, label='Forecast', color='red')
plt.fill_between(forecast_ci.index,
                 forecast_ci.iloc[:, 0],
                 forecast_ci.iloc[:, 1], color='pink', alpha=0.3)

plt.title("SARIMA Forecast - AirPassengers Dataset")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.legend()
plt.show()





```

```python
# 9. Apply Local Outlier Factor (LOF) on any numerical dataset to detect
# anomalies and visualize them using matplotlib.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import LocalOutlierFactor

# Generate synthetic numerical dataset
np.random.seed(42)
X_inliers = 0.3 * np.random.randn(100, 2)   # cluster 1
X_inliers = np.r_[X_inliers + 2, X_inliers - 2]  # shift to make 2 clusters

# Add anomalies (outliers far from clusters)
X_outliers = np.random.uniform(low=-6, high=6, size=(20, 2))

# Combine data
X = np.r_[X_inliers, X_outliers]

# Apply Local Outlier Factor
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.1)
y_pred = lof.fit_predict(X)  # -1 = anomaly, 1 = normal

# Separate normal and anomalies
normal = X[y_pred == 1]
anomalies = X[y_pred == -1]

# Plot results
plt.figure(figsize=(8,6))
plt.scatter(normal[:, 0], normal[:, 1], c='blue', label='Normal', alpha=0.7)
plt.scatter(anomalies[:, 0], anomalies[:, 1], c='red', marker='x', s=100, label='Anomaly')
plt.title("Local Outlier Factor (LOF) Anomaly Detection")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()

```

Question 10: You are working as a data scientist for a power grid monitoring company.
Your goal is to forecast energy demand and also detect abnormal spikes or drops in
real-time consumption data collected every 15 minutes. The dataset includes features
like timestamp, region, weather conditions, and energy usage.

Explain your real-time data science workflow:

● How would you detect anomalies in this streaming data (Isolation Forest / LOF /
DBSCAN)?

● Which time series model would you use for short-term forecasting (ARIMA /
SARIMA / SARIMAX)?

● How would you validate and monitor the performance over time?

● How would this solution help business decisions or operations?



Ans:Real-Time Data Science Workflow
1. Anomaly Detection in Streaming Data

 - Challenge: Detect abnormal spikes/drops in energy usage every 15 minutes.

--Options:

 - Isolation Forest → Good for high-dimensional streaming data (usage + weather + region). Scalable, fast.

 - Local Outlier Factor (LOF) → Detects local/contextual anomalies (e.g., a sudden spike in one region vs. others). More costly to compute in real-time, but useful if anomalies are regional/contextual.

 - DBSCAN → Better for batch/offline clustering; not ideal for real-time streams.

 - Choice: Isolation Forest for real-time anomaly detection (fast, handles large-scale data) + optional LOF in smaller subsets (per region).

2. Short-Term Forecasting Model

--Data Characteristics:

 - Collected every 15 minutes → strong seasonality (daily cycle, weekly cycle).

 - Influenced by weather conditions (temperature, humidity, etc.).

--Model Choices:

 - ARIMA: Handles trend but not seasonality well.

 - SARIMA: Handles trend + seasonality (daily/weekly patterns).

 - SARIMAX: SARIMA + external regressors (weather, holidays, regional demand).

  - Choice: SARIMAX — because external factors like temperature, humidity, region load directly impact demand.

3. Validation & Monitoring

--Offline Validation:

 - Train/test split (rolling window approach).

 - Metrics: RMSE / MAE for forecasts, Precision/Recall for anomaly detection.

--Real-Time Monitoring:

 - Continuously update model with sliding windows.

 - Drift detection: Monitor if forecast errors suddenly increase (concept drift).

 - Re-train models weekly/monthly as new seasonal patterns emerge.

4. Business Impact

---Operational Benefits:

 - Early warning system → Detect abnormal spikes/drops in consumption (equipment failure, sudden regional demand surge).

 - Load balancing → Forecast short-term demand to avoid overloads and blackouts.

 - Energy purchasing decisions → Accurate forecasts reduce costs (e.g., buying electricity in advance vs. peak prices).

 - Improved reliability → Better customer satisfaction and regulatory compliance.

-- Summary Workflow

 - Stream Ingestion: Collect energy usage + weather data every 15 minutes.

 - Anomaly Detection: Use Isolation Forest in real-time → flag spikes/drops.

 - Forecasting: Train SARIMAX (with weather + region as features) for 1–6 hour forecasts.

 - Validation: Use rolling forecasts, RMSE, Precision/Recall. Monitor drift.

 - Business Use: Enable proactive grid management, cost optimization, and outage prevention.
