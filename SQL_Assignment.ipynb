2. Explain the purpose of constraints and how they help maintain data integrity in a database. Provide examples of common types of constraints.?


Ans:Constraints in a database are rules applied to columns or tables to limit the type of data that can be inserted, updated, or deleted. Their primary purpose is to maintain data integrity, ensuring the accuracy, consistency, and reliability of the information stored within the database. By enforcing specific conditions, constraints prevent invalid or inconsistent data from entering the system, upholding defined business rules and relationships between data.

---How Constraints Maintain Data Integrity:

 - Preventing Invalid Data: Constraints validate data at the point of entry or modification, rejecting any data that violates the defined rules. This prevents erroneous or nonsensical values from corrupting the dataset.

 - Ensuring Consistency: Constraints help maintain consistency across related tables by enforcing referential integrity, ensuring that relationships between data remain valid.

 - Enforcing Business Rules: They allow the implementation of business logic directly within the database schema, ensuring that data adheres to real-world requirements (e.g., age cannot be negative).

 - Reducing Errors: By automating data validation, constraints reduce the likelihood of human error or application bugs introducing corrupt data.

---Examples of Common Types of Constraints:

 - 1. :NOT NULL: Ensures that a column cannot contain a NULL value, meaning a value must always be provided for that column.

     CREATE TABLE Employees (

        EmployeeID INT PRIMARY KEY,
        FirstName VARCHAR(50) NOT NULL

    );

  - 2.UNIQUE: Ensures that all values in a specific column or set of columns are unique across all rows in the table.
      CREATE TABLE Users (
        UserID INT PRIMARY KEY,
        Email VARCHAR(100) UNIQUE
    );

 - 3. PRIMARY KEY: Uniquely identifies each row in a table. It is a combination of NOT NULL and UNIQUE. A table can have only one primary key.

     CREATE TABLE Products (

        ProductID INT PRIMARY KEY,
        ProductName VARCHAR(100)
    );

 - 4.FOREIGN KEY: Establishes and enforces a link between data in two tables. It references the primary key of another table, ensuring referential integrity.

       CREATE TABLE Orders (

        OrderID INT PRIMARY KEY,
        ProductID INT,
        FOREIGN KEY (ProductID) REFERENCES Products(ProductID)
    );
 - 5.CHECK: Enforces a condition or rule on the values in a column. Only values that satisfy the condition are allowed.

         CREATE TABLE Accounts (

        AccountID INT PRIMARY KEY,
        Balance DECIMAL(10, 2) CHECK (Balance >= 0)
    );
 - 6. DEFAULT: Provides a default value for a column when no value is explicitly specified during insertion.
         CREATE TABLE Tasks (

        TaskID INT PRIMARY KEY,
        Status VARCHAR(20) DEFAULT 'Pending'
    );

    

3. Why would you apply the NOT NULL constraint to a column? Can a primary key contain NULL values? Justify
your answer.?

Ans:1. Why apply the NOT NULL constraint to a column:

 - The NOT NULL constraint ensures that a column cannot have a NULL value. This is important because:

 - It enforces data integrity by making sure that essential information is always provided.

 - Example: In an employees table, emp_name should not be NULL; every employee must have a name.

 - It prevents accidental omission of critical data during insert or update operations.

 - It can improve query performance since the database knows there will always be a value in that column.

Example sql:

CREATE TABLE employees (

    emp_id INT NOT NULL,
    emp_name VARCHAR(50) NOT NULL,
    age INT
);

Here, emp_id and emp_name cannot be left empty (NULL).


2. Can a primary key contain NULL values:

 - No, a primary key cannot contain NULL values.

 ---Justification:

 - A primary key uniquely identifies each row in a table.

 - If a primary key were allowed to be NULL, it would break uniqueness, because NULL represents an unknown or missing value.

 - Most databases automatically apply NOT NULL to primary key columns even if you don’t explicitly specify it.

 Example sql:


 CREATE TABLE employees (

    emp_id INT PRIMARY KEY,  -- automatically NOT NULL
    emp_name VARCHAR(50) NOT NULL
);

Trying to insert a row with emp_id = NULL will result in an error.

---Why apply a NOT NULL constraint:

 - Enforce data integrity: To ensure that every record has a value in a particular column, preventing incomplete or malformed data.

 - Mandatory fields: It makes certain fields, like an employee's hire date or a customer's unique ID, required for a record to be valid.

 - Improve query performance: In some cases, NOT NULL constraints can improve query performance by allowing the database system to optimize data access and indexing.

---Can a primary key contain NULL values:

 - No, a primary key column cannot contain NULL values.

---Justification:

 - Uniqueness: The primary key's core function is to provide a unique identifier for each row. A NULL value cannot be unique because it represents the absence of a value.

 - Uncomparable values: NULL values are not equal to any other value, including other NULLs. If a primary key were NULL, it couldn't be used to reliably distinguish one row from another.

 - Implicit NOT NULL: When you define a primary key on a column, most database systems automatically set that column to be NOT NULL, effectively making it a requirement that a value must be present.

The NOT NULL constraint ensures data is present, while the primary key ensures that the data uniquely identifies a record, making NULLs impermissible in a primary key.


4. Explain the steps and SQL commands used to add or remove constraints on an existing table. Provide an
example for both adding and removing a constraint.?

Ans:--1. Adding Constraints to an Existing Table:

You can use the ALTER TABLE statement to add constraints to a column or table after it has been created.

 - Syntax:

ALTER TABLE table_name
ADD CONSTRAINT constraint_name constraint_type (column_name);


---Common constraint types:

  - NOT NULL

 - UNIQUE

 - PRIMARY KEY

 - FOREIGN KEY

 - CHECK

 - Example: Adding a UNIQUE constraint

Suppose we have this employees table:

CREATE TABLE employees (

    emp_id INT,
    emp_name VARCHAR(50),
    email VARCHAR(100)
);


 - We want to ensure that email is unique.

ALTER TABLE employees
ADD CONSTRAINT unique_email UNIQUE (email);


-- Now, no two employees can have the same email.

 - Example: Adding a CHECK constraint

 - Ensure age is at least 18:

ALTER TABLE employees
ADD CONSTRAINT chk_age CHECK (age >= 18);

--2. Removing Constraints from an Existing Table

 - To remove a constraint, you again use ALTER TABLE with DROP CONSTRAINT.

 - Syntax:

ALTER TABLE table_name
DROP CONSTRAINT constraint_name;


 - Note: The exact syntax may vary slightly depending on the database (e.g., MySQL uses DROP INDEX for UNIQUE, DROP FOREIGN KEY for foreign keys, etc.).

 - Example: Removing the UNIQUE constraint

ALTER TABLE employees
DROP CONSTRAINT unique_email;


 - Example: Removing the CHECK constraint

ALTER TABLE employees
DROP CONSTRAINT chk_age;

 - Important Notes

Always give a name to constraints when adding them. This makes it easier to drop them later.

If you don’t name them, the database generates a system-defined name, which can be hard to reference.

 - NOT NULL is slightly different:

1. To add NOT NULL:

ALTER TABLE employees
MODIFY emp_name VARCHAR(50) NOT NULL;  -- MySQL


2. To remove NOT NULL:

ALTER TABLE employees
MODIFY emp_name VARCHAR(50) NULL;  -- MySQL

5.  Explain the consequences of attempting to insert, update, or delete data in a way that violates constraints.
Provide an example of an error message that might occur when violating a constraint?

Ans:Attempting to insert, update, or delete data in a way that violates database constraints results in the rejection of the operation and the generation of an error message. The primary consequence is the prevention of invalid or inconsistent data from entering or remaining in the database, thereby maintaining data integrity.

---Specifically, the consequences depend on the type of constraint violated:

 - Primary Key/Unique Constraint Violation: If an insertion or update attempts to create a duplicate value in a column or set of columns designated as unique or a primary key, the operation will be rejected. This ensures that each record can be uniquely identified.

--- Foreign Key Constraint Violation (Referential Integrity):

  - Insertion/Update: If an insertion or update attempts to reference a non-existent primary key in a related table, the operation is rejected, preventing "orphan" records.

 - Deletion/Update on Primary Key: If a deletion or update on a primary key attempts to remove or change a value that is still referenced by a foreign key in another table, the operation may be rejected (default behavior), or a cascade action might be triggered (e.g., cascading deletion of dependent records).

 - Check Constraint Violation: If an insertion or update attempts to store a value that does not satisfy a defined check condition (e.g., an age less than zero), the operation is rejected.

 - Not Null Constraint Violation: If an insertion or update attempts to store a NULL value in a column designated as NOT NULL, the operation is rejected.

 2. Examples of Constraint Violations and Error Messages:

 - a) NOT NULL Violation:

CREATE TABLE employees (

    emp_id INT PRIMARY KEY,
    emp_name VARCHAR(50) NOT NULL
);

-- Attempt to insert NULL in emp_name
INSERT INTO employees (emp_id, emp_name) VALUES (1, NULL);


Error Message (MySQL example):

ERROR 1048: Column 'emp_name' cannot be null

 - b) UNIQUE / PRIMARY KEY Violation:

CREATE TABLE employees (

    emp_id INT PRIMARY KEY,
    email VARCHAR(50) UNIQUE
);

-- Insert duplicate emp_id
INSERT INTO employees (emp_id, email) VALUES (1, 'a@example.com');
INSERT INTO employees (emp_id, email) VALUES (1, 'b@example.com');


Error Message:

ERROR 1062: Duplicate entry '1' for key 'PRIMARY'

 - c) CHECK Constraint Violation:

CREATE TABLE employees (

    emp_id INT,
    age INT CHECK (age >= 18)
);

-- Attempt to insert invalid age
INSERT INTO employees (emp_id, age) VALUES (2, 15);


Error Message:

ERROR: CHECK constraint failed: age >= 18

 - d) FOREIGN KEY Violation:

CREATE TABLE departments (

    dept_id INT PRIMARY KEY
);

CREATE TABLE employees (

    emp_id INT PRIMARY KEY,
    dept_id INT,
    FOREIGN KEY (dept_id) REFERENCES departments(dept_id)
);

-- Attempt to insert a non-existent department
INSERT INTO employees (emp_id, dept_id) VALUES (1, 10);


Error Message:

ERROR 1452: Cannot add or update a child row: a foreign key constraint fails


---Summary:

 - Violating constraints prevents invalid or inconsistent data from entering the database.

 - Common violations include NOT NULL, UNIQUE/PRIMARY KEY, CHECK, and FOREIGN KEY.

 - The database throws an error message indicating which constraint was violated.

1. -Identify the primary keys and foreign keys in maven movies db. Discuss the differences?

Ans:The Maven Movies database is a sample relational database (similar to Sakila, the DVD rental database) often used for SQL practice. Let’s break down the primary keys (PKs) and foreign keys (FKs) in the schema, and then discuss their differences.

--- Primary Keys (PKs):

A primary key uniquely identifies each row in a table.

It cannot contain NULL values and must be unique.

In Maven Movies, examples include:

 - actor → actor_id (PK)

 - address → address_id (PK)

 - category → category_id (PK)

 - city → city_id (PK)

 - country → country_id (PK)

 - customer → customer_id (PK)

 - film → film_id (PK)

 - film_actor → (film_id, actor_id) (Composite PK)

 - film_category → (film_id, category_id) (Composite PK)

 - inventory → inventory_id (PK)

 - language → language_id (PK)

 - payment → payment_id (PK)

 - rental → rental_id (PK)

 - staff → staff_id (PK)

 - store → store_id (PK)

---- Foreign Keys (FKs):

 - A foreign key is a column (or set of columns) that creates a link between data in two tables.

It references a primary key in another table.

Examples in Maven Movies:

 - address.city_id → references city.city_id

 - city.country_id → references country.country_id

 - customer.store_id → references store.store_id

 - customer.address_id → references address.address_id

 - film.language_id → references language.language_id

 - film.original_language_id → references language.language_id

 - film_actor.film_id → references film.film_id

 - film_actor.actor_id → references actor.actor_id

 - film_category.film_id → references film.film_id

 - film_category.category_id → references category.category_id

 - inventory.film_id → references film.film_id

 - inventory.store_id → references store.store_id

 - payment.customer_id → references customer.customer_id

 - payment.staff_id → references staff.staff_id

 - payment.rental_id → references rental.rental_id

 - rental.inventory_id → references inventory.inventory_id

 - rental.customer_id → references customer.customer_id

 - rental.staff_id → references staff.staff_id

 - staff.address_id → references address.address_id

 - staff.store_id → references store.store_id

 - store.manager_staff_id → references staff.staff_id

 - store.address_id → references address.address_id.



 | **Aspect**                    | **Primary Key (PK)**                                                            | **Foreign Key (FK)**                                                                       |
| ----------------------------- | ------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------ |
| **Definition**                | A column (or set of columns) that uniquely identifies each record in a table.   | A column (or set of columns) in one table that refers to the primary key in another table. |
| **Uniqueness**                | Must have unique values for every row.                                          | Can have duplicate values.                                                                 |
| **Null values**               | Cannot contain NULL values.                                                     | Can contain NULL values (unless specified as NOT NULL).                                    |
| **Number allowed**            | Only one primary key per table (but it can be composite with multiple columns). | A table can have multiple foreign keys.                                                    |
| **Purpose**                   | Ensures **entity integrity** (uniqueness of rows).                              | Ensures **referential integrity** (valid relationship between tables).                     |
| **Example (Maven Movies DB)** | `customer.customer_id` uniquely identifies each customer.                       | `payment.customer_id` refers to `customer.customer_id`.                                    |

1. First Normal Form (1NF):

A table is in 1NF if:

 - Each column contains atomic (indivisible) values.

 - Each row is unique.

 - There are no repeating groups or arrays in a single column.

 - Step 1: Identify a possible violation in Sakila

In the Sakila database, most tables are already normalized. However, suppose we hypothetically had a table like this:

| 2            | Alice Brown | 345-678          | [alice@gmail.com](mailto:alice@gmail.com)                                    |


 - Violation:

 - The columns phone_numbers and emails contain multiple values in a single cell, which violates 1NF.

 - Step 2: How to normalize it to 1NF

We need to ensure atomic values in each column. Create separate rows for each phone number and email:

 - Normalized Tables:

 - Customer Table (atomic info about the customer).

 | customer\_id | name        |
| ------------ | ----------- |
| 1            | John Smith  |
| 2            | Alice Brown |

| customer\_id | phone\_number |

| 2            | [alice@gmail.com](mailto:alice@gmail.com) |

Now, each column contains atomic values, and there are no repeating groups, so the tables satisfy 1NF.

2.  Second Normal Form (2NF):

      a. Choose a table in Sakila and describe how you would determine whether it is in 2NF.
     If it violates 2NF, explain the steps to normalize it.

     Ans:Second Normal Form (2NF)

A table is in 2NF if:

 - It is already in 1NF.

 - All non-key attributes are fully functionally dependent on the entire primary key, not just part of a composite key.

 - Violation: If a table has a composite primary key and some columns depend only on part of that key, it violates 2NF.

 - Step 1: Choose a Sakila table

 - A good example is the film_actor table:

 | actor\_id | film\_id | last\_update |
| --------- | -------- | ------------ |

 - Primary key: (actor_id, film_id) (composite key)

 - last_update is the non-key attribute.

 Step 2: Check for 2NF violation

The table is in 1NF (atomic values, no repeating groups). ✅

Check functional dependency:

last_update only depends on the combination (actor_id, film_id)?

Actually, last_update may not depend on the combination; it might be just the last time the actor or film record was updated, so it could depend on actor_id or film_id individually.

If last_update depends only on actor_id (or film_id), it violates 2NF because it is partially dependent on the composite key.

Step 3: Normalize to 2NF

Step 3a: Remove partially dependent columns to a new table.

Original film_actor Table (violates 2NF)

| actor_id | film_id | last_update |

Normalized film_actor Table (2NF) – only composite key columns:

| actor_id | film_id |

Separate actor_last_update Table – attributes depending on a single key:

| actor_id | last_update |

 - Now, all non-key attributes are fully dependent on the whole primary key.

 - Summary:

 - Check 2NF: Look for composite primary keys → see if any non-key column depends only on part of the key.

 - Normalize: Move partially dependent attributes to a separate table.

3. Third Normal Form (3NF):

   present and outline the steps to normalize the table to 3NF.?

 Ans:Third Normal Form (3NF)

A table is in 3NF if:

It is already in 2NF.

No transitive dependencies exist — that is, non-key attributes cannot depend on other non-key attributes.

 - Violation: If a non-key column depends on another non-key column instead of the primary key, the table violates 3NF.

No transitive dependencies remain in rental.

-  Summary

 - Normalize: Move transitive dependent attributes to the table where their direct key dependency exists.
