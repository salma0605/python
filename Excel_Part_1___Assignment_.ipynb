1.Question 1: What is the difference between 'Paste' and 'Paste Special' in Excel? Briefly
explain with examples. ?

Ans:In Excel, "Paste" inserts copied data with all its original formatting, formulas, and properties, while "Paste Special" allows you to choose exactly what to paste, offering options like values, formulas, formatting, comments, or a combination, providing greater control over how the data is placed in the destination. For example, a simple "Paste" of a cell with a formula and formatting will paste both, whereas "Paste Special" > "Values" will paste only the resulting number, stripping the formula.

1. Paste in Excel:

  - Definition: The regular "Paste" command copies everything from the copied cell(s): values, formulas, formatting, and comments.

 - Example:
If cell A1 contains the formula =B1+C1 and is formatted in bold blue, then:

 - Copy A1 → Paste into D1

 - D1 will have:

   - The formula =B1+C1

   - The bold blue formatting.

 2. Paste Special in Excel:

 - Definition: "Paste Special" gives you options to paste only specific parts of the copied data (like just values, just formulas, just formatting, etc.).

 - Examples:

 1. Paste Values Only

 - Copy A1 (which has =B1+C1, and suppose B1=5, C1=3 → result is 8).

 - Paste Special → Values into D1

 - D1 will show 8 (the result), not the formula.

2. Paste Formats Only

 - Copy A1 (bold blue text).

 - Paste Special → Formats into D1

 - D1 will keep its original value, but now the text will turn bold blue.

3. Paste Formulas Only

 - Copy A1 (=B1+C1).

 - Paste Special → Formulas into D1

 - D1 will have the formula =B1+C1 but not the formatting.


  - Paste Values: Copies only the results of formulas, not the formulas themselves. If you copy a cell with a formula =A1+1 and paste its value into a new cell, the new cell will contain the number that resulted from the formula, not the formula itself.

 - Paste Formatting: Copies only the visual styles and formatting (like font, colors, and borders) but not the content or formulas.

 - Transpose: Switches the orientation of the data, converting rows into columns and vice versa.

 - Paste Link: Creates a dynamic link between the source and destination, so changes to the original data will automatically update the pasted data.

2. Question 2: Describe the functions and usefulness of 'Freeze Panes' and 'Split Panes'
in Excel. ?

Ans:Freeze Panes and Split Panes are Excel features designed to improve navigation and viewing of large spreadsheets by controlling which parts of the worksheet are visible while scrolling. While they both involve dividing the screen, they differ significantly in function and application.

1. Freeze Panes:

 - Function
Freeze Panes locks one or more specific rows or columns in place so they remain visible at all times, regardless of where you scroll in the worksheet.

 - Freeze Top Row: Locks the very first row, which typically contains column headers.

 - Freeze First Column: Locks the first column, which is useful for keeping row labels in view.

 - Freeze Multiple Panes: By selecting a specific cell (e.g., C3), you can freeze all rows above that cell (rows 1 and 2) and all columns to the left of that cell (columns A and B).

----Usefulness:

 - Data entry: When working with a large dataset, you can scroll down thousands of rows and still have the column headers (such as "Name," "Product ID," or "Date") visible, ensuring you enter data into the correct fields.

 - Data analysis: It helps you keep important context on screen, such as a product's name or a customer's ID, as you scroll to different columns to analyze their sales figures or other metrics.

 - Presentation: When presenting a large spreadsheet, Freeze Panes keeps the critical header information on display for the audience, even as you navigate to different data sections.

2. Split Panes:

 - Function
Split Panes divides your worksheet window into two or four separate, scrollable panes. Each pane functions as its own independent window, but they all display the same worksheet. You can scroll and view different parts of the same worksheet simultaneously.

 - Create panes: You can create a horizontal split by selecting a row, a vertical split by selecting a column, or a four-way split by selecting a cell and then clicking the Split button on the View tab.

 - Independent scrolling: Each pane has its own set of scroll bars, allowing you to view distant sections of the same sheet side-by-side.

 - Adjustable: The dividing bars can be clicked and dragged to resize the panes.

 ---- Usefulness:

 - Compare data: Split Panes is ideal for comparing data from widely separated sections of a worksheet. For instance, you could compare a total at the bottom of a sheet with data at the top, or compare the sales figures for January (column B) with those for December (column M) without scrolling back and forth.

 - Simultaneous view: It allows you to enter data in one pane while simultaneously monitoring the effects or comparisons in another pane.

 - Local focus: If you need to focus on a specific area while keeping another distant section visible for reference, Split Panes is the more flexible tool.



 ---Key Difference:

 - Freeze Panes → Keeps specific rows/columns visible at all times.

 - Split Panes → Divides the window into sections so you can scroll each independently.

3. Question 3: Explain the difference between inserting a new row and inserting a new
column in Excel. Can you insert multiple rows or columns at once?

Ans:Inserting a new row adds data vertically from top to bottom, while inserting a new column adds data horizontally from left to right. Yes, you can insert multiple rows or columns at once by selecting the desired number of existing rows or columns before performing the insert operation. For rows, select the number of rows above where you want to insert, and for columns, select the number of columns to the left of where you want to insert, then use the right-click menu or the Insert option on the Home tab.  

1. Inserting a New Row:

 - What happens: A blank row is added above the selected row.

 - Effect: All existing rows shift down by one.

 - Example:
If you select row 3 and insert a new row, the new blank row will appear as Row 3, and the old Row 3 becomes Row 4.

 - Purpose: Adds a new horizontal line to accommodate more data entries, shifting existing rows downward.

 - Method:
 - Select any cell within the row above where you want to insert a new row.

 - Right-click on the selected row.

 - From the context menu, choose Insert and then select Table Rows Above or a similar option.

 - Alternatively, go to the Home tab, click the Insert dropdown in the Cells group, and choose Insert Sheet Rows.

2. Inserting a New Column:

 - What happens: A blank column is added to the left of the selected column.

 - Effect: All existing columns shift right by one.

 - Example:If you select column C and insert a new column, the new blank column will appear as Column C, and the old Column C becomes Column D.

  - Purpose: Adds a new vertical line for more data attributes, shifting existing columns to the right.

 - Method:

 - Select any cell within the column to the left of where you want the new column to appear.

 - Right-click on the selected column.

 - From the context menu, choose Insert and then select Table Columns to the Left or a similar option.

 - Alternatively, go to the Home tab, click the Insert dropdown in the Cells group, and choose Insert Sheet Columns.

3. Can You Insert Multiple Rows or Columns at Once:

 - Rows: Select as many existing rows as you want to insert, then choose Insert.

 - Example: Select 3 rows (Rows 4–6) → Right-click → Insert → Excel will insert 3 new rows above Row 4.

 - Columns: Select as many existing columns as you want to insert, then choose Insert.

 -  Example: Select 2 columns (Columns B–C) → Right-click → Insert → Excel will insert 2 new blank columns before Column B.


  ---- In short:

 - Insert Row → adds blank row(s) above selection, shifts data down.

 - Insert Column → adds blank column(s) to the left, shifts data right.

 - Multiple inserts are possible by selecting multiple rows/columns first.

4. Question 4: What are logical functions in Excel? Provide examples of at least two
logical functions and their applications.?

Ans:Logical functions in Excel test conditions to perform actions or return TRUE/FALSE values, helping to make decisions within spreadsheets. Key examples include the IF function, which returns one value if a condition is TRUE and another if it's FALSE, and the AND function, which returns TRUE only if all specified conditions are met. These functions automate decision-making by creating conditional formulas that respond dynamically to data, enabling tasks like conditional formatting, data validation, and workflow automation.


---Logical Functions in Excel:

 - Definition: Logical functions are used to test conditions (TRUE/FALSE) and return results based on those conditions.

 - Usefulness: They help in decision-making, data analysis, and applying conditions to formulas.


 ---Examples of Logical Functions:

1. IF Function

 - Syntax: =IF(logical_test, value_if_true, value_if_false)

 - Purpose: Returns one value if the condition is TRUE and another if FALSE.

 - Example:

 - Formula: =IF(A1>=50, "Pass", "Fail")

 - Meaning: If the value in cell A1 is 50 or more, it shows “Pass”, otherwise “Fail”.

 - Application: Useful in grading systems, sales targets, and performance evaluations.


 2. AND Function:


 - Syntax: =AND(condition1, condition2, …)

 - Purpose: Returns TRUE only if all conditions are TRUE.

 - Example:

 - Formula: =AND(A1>=50, B1>=50)

 - Meaning: Returns TRUE only if both A1 and B1 are greater than or equal to 50.

 - Application: Used when multiple criteria must be met (e.g., eligibility checks).

 3. OR Function (bonus example):

 - Syntax: =OR(condition1, condition2, …)

 - Purpose: Returns TRUE if any condition is TRUE.

 - Example:

 - Formula: =OR(A1>=50, B1>=50)

 - Meaning: Returns TRUE if either A1 or B1 is greater than or equal to 50.

 - Application: Useful when passing only requires one subject score above 50.


 ---Other Logical Functions:

 - OR Function: Returns TRUE if at least one of the specified conditions is TRUE.

 - NOT Function: Returns the opposite of the logical value it is given (e.g., if the input is TRUE, it returns FALSE).

 - XOR Function: Returns TRUE if an odd number of arguments are TRUE.

 - IFS Function: Evaluates multiple conditions and returns a value for the first condition that evaluates to TRUE.


5. Question 5: Discuss the purpose of 'XLOOKUP' and how it differs from the traditional
'VLOOKUP' function.?

Ans:The purpose of the XLOOKUP function is to search for a value in a range or array and return a corresponding item from a different range. Introduced in 2019, it is designed as a more flexible and robust replacement for older lookup functions like VLOOKUP and HLOOKUP, addressing many of their common limitations.

| Feature                    | **VLOOKUP**                                                             | **XLOOKUP**                                                                                  |
| -------------------------- | ----------------------------------------------------------------------- | -------------------------------------------------------------------------------------------- |
| **Lookup Direction**       | Searches **only vertically** (top to bottom).                           | Searches **both vertically and horizontally**.                                               |
| **Column Selection**       | Needs a **column index number** (e.g., 2 means return from 2nd column). | No column index needed — just provide the return range.                                      |
| **Exact Match Default**    | Default is **approximate match**, unless you specify `FALSE`.           | Default is **exact match**, making it safer.                                                 |
| **Lookup Flexibility**     | Can only search **left to right**.                                      | Can search **left to right** or **right to left**.                                           |
| **Return Multiple Values** | Returns only **one column** of data.                                    | Can return **multiple columns/rows** at once.                                                |
| **Error Handling**         | Shows `#N/A` if not found.                                              | Has a built-in **if_not_found** argument (you can customize the message, e.g., “Not Found”). |


Examples
✅ VLOOKUP Example

Suppose you have a table:

| ID  | Name | Marks |
| --- | ---- | ----- |
| 101 | Ali  | 85    |
| 102 | Riya | 90    |
| 103 | John | 70    |

 - Formula:

=VLOOKUP(102, A2:C4, 2, FALSE)

Looks for 102 in column A

Returns from column 2 → “Riya”

---XLOOKUP Example:


 - Same table, but formula:
=XLOOKUP(102, A2:A4, B2:B4, "Not Found")

 - Looks for 102 in column A

 - Returns the matching value from column B → “Riya”

 - If not found → returns “Not Found”

 - Also, you could look backward:
=XLOOKUP("Riya", B2:B4, A2:A4) → returns 102

7. Question 7: Demonstrate how to insert and delete multiple rows and columns in Excel.?

Ans:To insert or delete multiple rows and columns in Excel, you first need to select the desired number of rows or columns, and then use a right-click menu, the Ribbon, or a keyboard shortcut.

1. How to insert multiple rows:

---Method 1: Using the right-click menu:

 - Select the rows. Click and drag your mouse across the row numbers to select the same number of rows you want to insert. The new blank rows will appear above your selection.

 - Right-click. Right-click one of the selected row numbers.

 - Click Insert. Select Insert from the context menu that appears.

----Method 2: Using the Ribbon:

 - Select the rows. Drag your mouse over the row numbers to select the number of rows you want to add.

 - Click Home. Go to the Home tab on the Ribbon.
Click Insert. In the Cells group, click the dropdown arrow next to Insert, and then choose Insert Sheet Rows.

----Method 3: Using a keyboard shortcut:

 - Select the rows. Click on a cell in the row where you want to add new ones. Press Shift + Spacebar to select the entire row.

 - Select additional rows. While holding the Shift key, use the down arrow key to select the number of additional rows you want to insert.

 - Insert the rows. Press Ctrl + Shift + + (the plus sign) to add the new rows.

2. How to delete multiple rows:

----Method 1: Using the right-click menu:

 - Select the rows. Click and drag your mouse across the row numbers to select the rows you want to delete. For non-adjacent rows, hold down the Ctrl key while clicking each row number.

 - Right-click. Right-click one of the selected row numbers.

 - Click Delete. Select Delete from the context menu.

 ---Method 2: Using the Ribbon:

 - Select the rows. Drag your mouse across the row numbers you want to remove.

 - Click Home. Navigate to the Home tab.
Click Delete. In the Cells group, click the dropdown arrow next to Delete, and then choose

-   Delete Sheet Rows.

----Method 3: Using a keyboard shortcut:

 - Select the rows. Highlight the rows you want to delete by clicking and dragging across the row numbers.

 - Delete the rows. Press Ctrl + - (the minus sign) to remove the selected rows.

10. Question 10: You're working with a dataset that contains missing values. As a Data
Scientist, explain how you'd detect and handle missing data using Excel.

Mention tools like:

● Go To Special

● ISBLANK

● COUNTBLANK.

Ans:1. Detecting Missing Data in Excel:

a) Go To Special

 - Purpose: Quickly selects all blank cells in a dataset.

 - Steps:

 - Select the dataset.

 - Go to Home → Find & Select → Go To Special → Blanks.

 - All blank cells in your selection are highlighted.

 - Usefulness: Makes it easy to visually locate missing values and apply actions (like filling in values).

b) ISBLANK Function:


 - Purpose: Tests if a specific cell is empty (TRUE if blank, FALSE if not).

 - Syntax:

 - =ISBLANK(cell)


 - Example:

 - Dataset:

 - Name	Age
 - Ali	28
 - Riya
 - John	26

 - Formula: =ISBLANK(B2)

 - Returns FALSE (cell has 28)

 - Formula: =ISBLANK(B3)

 - Returns TRUE (cell is blank)

  c) COUNTBLANK Function

 - Purpose: Counts the total number of blank cells in a range.

 - Syntax:

 - =COUNTBLANK(range)


 - Example:

 - =COUNTBLANK(B2:B6)


 - Returns the number of missing values in the Age column.

2. Handling Missing Data in Excel

 - Once detected, you can handle missing data in several ways:

 1.  Manual Entry

 - Fill in missing values manually if known.

 2.  Use Fill Options

 - Select blank cells → Right-click → Fill → Series / Down / Up

 3.  Use Formulas

 - Replace blanks with default values using IF + ISBLANK:

 - =IF(ISBLANK(B2), 0, B2)


 - Fills blank cells with 0 (or any chosen value).

 4. Delete Missing Data

 - If appropriate, remove rows with blank cells:

 - Select the dataset → Home → Find & Select → Go To Special → Blanks → Delete → Entire Row

--- In short:

 - Detect: Go To Special → Blanks, ISBLANK, COUNTBLANK

 - Handle: Fill, replace with default, or remove missing values
